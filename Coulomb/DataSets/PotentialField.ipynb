```python
import os
import matplotlib.pyplot as plt
from PIL import Image, ImageChops
import numpy as np
```

```python
# Ruta a las carpetas de imágenes
acept_folder = "./acceptor"
donno_folder = "./donnor"
neutr_folder = "./neutral"

# Lista para almacenar las rutas de todas las imágenes
acept_paths = []
donno_paths = []
neutr_paths = []

# Recorrer las carpetas y almacenar las rutas de las imágenes
for filename in os.listdir(acept_folder):
    if filename.endswith(".png") or filename.endswith(".jpg"):
        acept_paths.append(os.path.join(acept_folder, filename))

for filename in os.listdir(donno_folder):
    if filename.endswith(".png") or filename.endswith(".jpg"):
        donno_paths.append(os.path.join(donno_folder, filename))    

for filename in os.listdir(neutr_folder):
    if filename.endswith(".png") or filename.endswith(".jpg"):
        neutr_paths.append(os.path.join(neutr_folder, filename))    
```

```python
print(f"Found {len(acept_paths)} acceptor images, {len(donno_paths)} donor images, and {len(neutr_paths)} neutral images.")
```

```python
# Hacer una matriz 4x4 de las imágenes de acceptor, con títulos solo para filas y columnas

figsize = (5.5, 5.5)
fig, axes = plt.subplots(4, 4, figsize=figsize)
fig.subplots_adjust(hspace=0.0, wspace=0.0)

acept_paths_sorted = sorted(acept_paths)

# Ejemplo de valores para E y B (ajusta según corresponda a tus imágenes)
E_values = [0, 1E3, 1E4, 1E5]  # Cambia por los valores reales de E
B_values = [0, 5, 10, 20]      # Cambia por los valores reales de B

for i in range(4):
    for j in range(4):
        idx = i * 4 + j
        ax = axes[i, j]
        if idx < len(acept_paths_sorted):
            img = Image.open(acept_paths_sorted[idx])
            ax.imshow(img)
        ax.axis('off')
        # Títulos de columnas (arriba)
        if i == 0:
            ax.set_title(f"$B$ = {B_values[j]} T", fontsize=7, pad=2)
        # Títulos de filas (al lado)
        if j == 0:
            ax.text(0, 0.5, f"$E$ = {E_values[i]} V/m", transform=ax.transAxes,
                    fontsize=7, va='center', ha='right', rotation=90)

plt.tight_layout(rect=[0.13, 0.13, 0.98, 0.95])
# plt.savefig("output/matrix_acceptor_images.pdf", format="pdf")
plt.show()
```

```python
# Hacer una matriz 4x4 de las imágenes de donnor, con títulos solo para filas y columnas

figsize = (5.5, 5.5)
fig, axes = plt.subplots(4, 4, figsize=figsize)
fig.subplots_adjust(hspace=0.0, wspace=0.0)

acept_paths_sorted = sorted(neutr_paths)

# Ejemplo de valores para E y B (ajusta según corresponda a tus imágenes)
E_values = [0, 0.01, 0.1, 1]  # Cambia por los valores reales de E
B_values = [0, 5, 10, 20]      # Cambia por los valores reales de B

# Crear un mapeo de colores para el colorbar
cmap = plt.get_cmap('turbo')
norm = plt.Normalize(vmin=0, vmax=60)

for i in range(4):
    for j in range(4):
        idx = i * 4 + j
        ax = axes[i, j]
        if idx < len(acept_paths_sorted):
            img = Image.open(acept_paths_sorted[idx])
            # Recortar 40 píxeles de la derecha
            img_cropped = img.crop((0, 60, img.width - 150, img.height))
            ax.imshow(img_cropped)
        ax.axis('off')
        # Títulos de columnas (arriba)
        if i == 0:
            ax.set_title(f"$B$ = {B_values[j]} T", fontsize=8, pad=2)
        # Títulos de filas (al lado)
        if j == 0:
            ax.text(0, 0.5, f"$E$ = {E_values[i]} kV/cm", transform=ax.transAxes,
                    fontsize=8, va='center', ha='right', rotation=90)

# Agregar colorbar a la figura completa
from matplotlib.cm import ScalarMappable
sm = ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])  # Necesario para fig.colorbar

# El colorbar se ubica fuera de la cuadrícula
cbar = fig.colorbar(sm, ax=axes, orientation='vertical', fraction=0.045, pad=0.02, location='right')
cbar.ax.tick_params(labelsize=7)
cbar.ax.set_title('mV', fontsize=8, pad=10)

#plt.tight_layout(rect=[0.13, 0.13, 0.98, 0.95])
plt.savefig("output/matrix_neutral_images_bar.pdf", format="pdf")
plt.show()
```
